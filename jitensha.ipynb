```python
import pandas as pd

df = pd.read_csv("jitensha.csv")
df.head()
```

```python
total = df.groupby('市区町村（発生地）').size().sum()
print(total)
```

```python
# 被害者の年齢ごとの件数
print("【被害者の年齢ごとの件数】")
print(df.groupby('被害者の年齢').size().sort_values(ascending=False))

# 空行で区切り
print("\n")

# 被害者の職業ごとの件数
print("【被害者の職業ごとの件数】")
print(df.groupby('被害者の職業').size().sort_values(ascending=False))
```
